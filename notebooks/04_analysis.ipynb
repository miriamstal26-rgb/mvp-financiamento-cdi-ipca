# 3. PERGUNTA 1

## Como CDI, Selic e IPCA evoluíram ao longo dos últimos dez anos?

Esta análise tem como objetivo comparar a evolução temporal dos três principais indicadores utilizados neste estudo.

A observação simultânea das séries históricas permite identificar tendências, mudanças de ciclo econômico e períodos de maior ou menor estabilidade econômica.

```python
display(
    gold_evolucao_indicadores.orderBy("ano_mes")
)
```

### Observação Metodológica

Os indicadores analisados possuem naturezas e periodicidades distintas.

CDI e Selic são representados por taxas anualizadas (% a.a.), enquanto o IPCA representa a variação percentual dos preços ao consumidor em base mensal (% a.m.).

Dessa forma, o objetivo desta análise não é comparar diretamente a magnitude absoluta dos valores observados, mas sim identificar tendências, ciclos econômicos e períodos de aceleração ou desaceleração de cada indicador ao longo do tempo.


### Exemplo de Interpretação Econômica

Em julho de 2022, observou-se um cenário caracterizado por taxas de juros elevadas, com CDI de 13,15% ao ano e Selic de 13,25% ao ano. No mesmo período, o IPCA registrou variação de -0,68% no mês, caracterizando um episódio de deflação.

Embora os valores não possam ser comparados diretamente devido às diferenças de periodicidade e escala dos indicadores, a combinação desses resultados fornece informações importantes sobre o contexto econômico do período.

A presença simultânea de juros elevados e inflação negativa sugere um ambiente de política monetária restritiva. Em outras palavras, o Banco Central mantinha taxas de juros elevadas como instrumento de controle inflacionário, mesmo após a desaceleração observada nos índices de preços.

Esse contexto pode ser interpretado como um período de elevada pressão monetária, situação em que o custo do crédito tende a permanecer elevado, reduzindo o consumo e os investimentos, enquanto a política econômica busca conter ou estabilizar a inflação.

Para consumidores com operações financeiras indexadas a juros de mercado, períodos caracterizados por CDI e Selic elevados tendem a representar momentos de maior pressão financeira, independentemente do comportamento pontual da inflação em um determinado mês.

# 3. PERGUNTA 4

## Em quais períodos um financiamento indexado ao CDI apresentaria maior pressão financeira para o consumidor?

Embora não tenha sido realizada uma simulação financeira direta, é possível identificar períodos de maior pressão financeira a partir do comportamento histórico do CDI.

Como o CDI representa uma referência amplamente utilizada pelo mercado financeiro, períodos caracterizados por taxas mais elevadas tendem a estar associados a maior custo do crédito, encarecimento das operações financeiras e aumento do esforço necessário para manutenção de financiamentos indexados a juros de mercado.

Para responder a essa questão, foi analisada a evolução histórica do CDI ao longo do período estudado.

```python
gold_evolucao_indicadores.select(
    "ano_mes",
    "cdi_percentual"
).orderBy(
    gold_evolucao_indicadores.cdi_percentual.desc()
).show(10, False)
```

```python
display(
    gold_evolucao_indicadores.select(
        "ano_mes",
        "cdi_percentual"
    ).orderBy("ano_mes")
)
```

Observa-se que o indicador atingiu seus menores patamares entre 2020 e 2021, período caracterizado por juros reduzidos e menor custo do crédito.



### Interpretação dos Resultados

A análise da evolução histórica do CDI permite identificar os períodos de maior pressão financeira para consumidores expostos a operações indexadas às taxas de juros de mercado.

Observa-se que o indicador atingiu seus menores patamares entre 2020 e 2021, período caracterizado por juros reduzidos e menor custo do crédito.

Esse comportamento esteve associado ao contexto econômico observado durante a pandemia de COVID-19. Diante da desaceleração da atividade econômica, foram adotadas medidas de estímulo monetário, incluindo a redução das taxas básicas de juros, com o objetivo de incentivar o consumo, o investimento e a recuperação econômica.

Como consequência, o CDI atingiu os menores níveis da série histórica analisada, contribuindo para um ambiente de crédito relativamente mais acessível quando comparado aos demais períodos do estudo.

A partir de 2021 ocorre um movimento de forte elevação das taxas de juros, refletindo uma política monetária mais restritiva adotada para controle da inflação.

Os maiores valores da série foram observados entre 2025 e 2026, quando o CDI permaneceu próximo de 14,90% ao ano. Em cenários como esse, financiamentos, empréstimos e demais operações financeiras indexadas aos juros tendem a apresentar custos mais elevados para consumidores e empresas.

# 5. APLICAÇÃO DOS RESULTADOS AO NEGÓCIO
Sob a perspectiva de uma cooperativa de crédito — segmento no qual a autora deste estudo se insere —, os resultados sugerem que o cenário mais recente da série histórica apresenta desafios relevantes para a expansão das carteiras de financiamento imobiliário. Entre janeiro de 2025 e junho de 2026, o CDI mensal evoluiu de 13,15% para 14,15% a.a., tendo atingido 14,90% a.a. em junho de 2025, o que caracteriza um dos patamares mais elevados de toda a série histórica analisada, comparável ao nível observado em 2016.

Esse cenário de juros elevados eleva o custo de captação das cooperativas de crédito, cujas fontes de recursos e produtos de crédito costumam estar referenciados, direta ou indiretamente, ao CDI. Como consequência, as taxas ofertadas nas linhas de financiamento imobiliário tendem a se elevar, tornando a contratação menos atrativa para o consumidor final e reduzindo o volume de novas operações — o que torna mais desafiador o crescimento dessa carteira específica.

Nesse contexto, estratégias voltadas exclusivamente para a ampliação da oferta de financiamentos tendem a apresentar resultados limitados. Ganha relevância, portanto, o fortalecimento do relacionamento com os cooperados, a ampliação da base de associados e a diversificação dos produtos financeiros oferecidos pela instituição. A pulverização da carteira — por meio do aumento do número de cooperados associado à oferta de soluções complementares de crédito, investimentos e seguros — contribui para reduzir a dependência de um único produto e para aumentar a resiliência da cooperativa diante de ciclos econômicos marcados por juros elevados.

Como limitação, cabe destacar que este estudo se restringiu ao período de julho de 2016 a junho de 2026 da série histórica, contemplando um ciclo completo de alta, queda e retomada da Selic, mas sem comparações com outras instituições do segmento, o que pode ser explorado em pesquisas futuras. Ainda assim, o monitoramento contínuo dos indicadores econômicos analisados pode auxiliar a cooperativa na definição de estratégias comerciais mais aderentes ao cenário vigente, permitindo equilibrar crescimento, competitividade e gestão de risco.
